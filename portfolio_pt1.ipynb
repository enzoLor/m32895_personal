### Ex. 6

```python
import random
print("Guess my Number!")

masterint = random.randint(1,10)
success = False
guesses = 0
currentGuess = 0

while success != True:
    currentGuess = int(input("Guess a number under 10."))
    guesses += 1
    if currentGuess > 10:
        print("I said not over 10.")
    if currentGuess == masterint:
        print("Well Done!! You got it in ", guesses, "guesses!")
        success = True
    if currentGuess < masterint:
        print("Yikes, too low.")
    if currentGuess > masterint and currentGuess < 10:
        print("Ouch, too high.")
```

```python
from math import factorial

def Maffs(x, N):
    return (x*5^2)/(1+factorial(2))+(x*25^2)/(2+factorial(3))+N

ix = int(input("Input x."))
iN = int(input("Input N."))

print("Your answer is ", Maffs(ix, iN))
```

```python
import numpy as np

tempData = np.array([22,24,19,21,23,25,26,27,24,23,20,22,21,19,18,22,23,24,26,28,30,27,25,26,24,22,21,20,19,23])

#Part 2
print("Minimum Temp (°C): ", np.min(tempData))
print("Maximum Temp (°C): ", np.max(tempData))
print(np.average(tempData))
print("")
#Part 3
tempDataF = ((9 * tempData / 5) + 32)
print("Temperature converted to Farenheit: ", tempDataF)
print("")
#Part 4
print("Days where temperature was above 25°C:", np.where(tempData > 25)[0])
print("")

#Part 5
print("Days where temperature was 20°C:", np.count_nonzero(tempData < 20))
print("")

#Part 6
print("Sorted list of days: ", np.sort(tempData))
print("")

#Part 7
print("Reshaped array: ")
print(np.reshape(tempData, (5, 6)))
```
