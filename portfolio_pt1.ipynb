```python
import random
print("Guess my Number!")

masterint = random.randint(1,10)
success = False
guesses = 0
currentGuess = 0

while success != True:
    currentGuess = int(input("Guess a number under 10."))
    guesses += 1
    if currentGuess > 10:
        print("I said not over 10.")
    if currentGuess == masterint:
        print("Well Done!! You got it in ", guesses, "guesses!")
        success = True
    if currentGuess < masterint:
        print("Yikes, too low.")
    if currentGuess > masterint and currentGuess < 10:
        print("Ouch, too high.")
```
